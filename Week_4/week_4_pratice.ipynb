### Attributes

- Attributes are variable associated with a class/object. They define the state or characteristics of an object.
- they are created as soon as class is instantiated, and they can be accessed using the dot notation.

There are two types:

1. `Instance Attributes`: Unique to each object, defined inside the `__init__` method using self.
2. `Class Attributes`: Shared by all objects of a class, defined directly under the class block

### Methods

Methods are function defined inside a class. They define the behavior of an object

what it can do.
- The first parameter in instance methods is always self.
- You call them using `object.method_name()`


### Reference: Varables
- Reference variables hold the object.
- We can create objects without reference variables, but we cannot access them later.
- An object can have multiple reference varibles
- Assigning a new reference variable to an object does not create a new object; it simply creates another reference to the same object. 

```python
class Person():
    def __init__(self):
        self.name = "Ankit"
        self.age = "25"

p = Person()
print(p.name)
q = p 
```

```python
print(p.age)
print(p.name)
q.name = "David"

print(q.name)
print(p.name)
```

- it is important to note that when we assign a new reference variable to an existing object, we are not creating a new object; instead, we are creating another reference to the same object in memory. This means that any changes made through one reference variable will affect the same object, and thus be reflected when accessed through any other reference variable that points to that object.

### Pass by reference

```python
class Person():
    def __init__(self, name, age):
        self.name = name
        self.age = age

# outside the class
def greet(person):
    print(f"Hello {person.name}, you are {person.age} years old.")
    

p = Person("Ankit", 25)
greet(p)
```

# Magic Methods or Dunder Methods

### what are Magic Methods?

Magic Methods are special built-in methods in Python that start and end with double iinderscores`(__like_this__)`. they are also known as dunder methids(shoert for "double undescore").

These methods are automatically called during specific operation like creating an object, printing it, adding two objects, comparing them, etc.

### Use of Magic methods

- to customize object behavior for built -in operations.
- to make your classes behave like built-in types(eg. how str() works on your object).
- to make debugging, logging, and operator overloading easier. 

**Magic Methods does not need to be called it run with the class same as `__init__`.**

```python
class book():
    def __init__(self, name_val, author_val):
        self.name = name_val
        self.author = author_val
    

    def my_method():
        print("this was called by the user\n")

my_book = book("LLM Fundemental", "Ankit")
print(my_book)

print("after applying dunder method\n")


### After appling the method
class book():
    def __init__(self, name_val, author_val):
        self.name = name_val
        self.author = author_val
    
    def __str__(self):  # Dunder method this to return when you say call objt 
        return f"{self.name} has author {self.author}"

    def my_method():
        print("this was called by the user\n")

my_book = book("LLM Fundemental", "Ankit")
print(my_book)
# I never called the magic method here
```

```python
class money():
    def __init__(self, amount_val):
        self.amount = amount_val

    def __str__(self):
        return f"RS: {self.amount}/-"
    

wallet_1 = money(100)
print(wallet_1)
wallet_2 = money(400)

print(wallet_1 + wallet_2)
```

```python
class money():
    def __init__(self, amount_val):
        self.amount = amount_val

    def __str__(self):
        return f"RS: {self.amount}/-"

    
    def __add__(s, o): # for convenction you can put __add__(self, other):
        return [s.amount + o.amount]  # now the number is printed in list
        # return money(self.amount + other.amount)  

wallet_1 = money(100)
wallet_2 = money(400)

print(wallet_1 + wallet_2)
```

### What is Encapsulation? 🔒

Encapsulation is the concept of hiding internal objects details and exposing only what is necessary. it helps in protecting data from unauthorized access and building data + methods that operate on the data in a single unit(i.e. a class).

|Syntax|Meaning
1. `name` --> Public attribute
2. `_name` --> Protected(internal use)
3. `__name` -->private(name mangling)




```python
class BankAccount:
    def __init__(self, owner, amount):
        self.owner = owner
        self.amount = amount

    def __str__(self):
        return f"Owner: {self.owner} has amount {self.amount}"
    
abc = BankAccount("Ankit", "1000")
abc.amount = 500000
print(abc)

```

### Getter and setter method
- this is based on the concept that a private attribute of a class cannot be accessed directly from outside the class. To access and modify these private attributes, we use getter and setter methods.
- Getter and setter methods are used to access and modify private attributes of a class.
- Getter method is used to retrieve the value of a private attribute, while setter method is used to set or update the value of a private attribute.

```python
# getter example
class BankAccount:
    def __init__(self, owner, amount):
        self.owner = owner
        self.__amount = amount  # private attribute

    def __str__(self):
        return f"Owner: {self.owner} has amount {self.__amount}"
    
    # getter method
    def get_amount(self):
        return self.__amount
    
    # setter method
    def set_amount(self, new_amount):
        if new_amount >= 0:
            self.__amount = new_amount
        else:
            print("Amount cannot be negative")
```

### why we need getter and setter method?
- the benefit of using getter and setter methods is that they provide a controlled way to access and modify private attributes. They allow you to add validation logic, perform additional actions when getting or setting a value, and maintain encapsulation by keeping the internal representation of the data hidden from the outside world.

```python
class BankAccount:
    def __init__(self, owner, amount):
        self.owner = owner
        self.amount = amount

    def __str__(self):
        return f"Owner: {self.owner} has amount {self.__amount}"
    
abc = BankAccount("Ankit", "1000")
abc.__amount = 500000
print(abc)

# when use __anyvar this tell that this is private data and do not need to ne changes
# name managgling abc_BankAccount__amount to acess it. 
```

```python
class BankAccount:
    def __init__(self, owner, amount):
        self.owner = owner
        self.__amount = float(amount)
        
    def deposit(self, amount):   ## getter 
        if amount > 0:
            self.__amount += amount
            print(f"Succesfully added {amount}")
        else:
            print("This was an invalid amount to added.")

    def withdraw(self, amount):    # getter
        if 0 < amount < self.__amount:
            self.__amount -= amount
            print(f"Succesfully withdrawn {amount}")
        else:
            print("This was an invalid amount to withdrawn")

    def get_balance(self):    #setter
        return self.__amount   


    def __str__(self):
        return f"Owner: {self.owner} has amount {self.__amount}"
    
abc = BankAccount("Ankit", "1000")
abc.deposit(5000)
print(abc.get_balance())
```

### Static method vs Instance method

- Static method is a method that belongs to the `class` rather than an instance of the class. It does not require an instance to be called and does not have access to the instance (self) or class (cls) variables. Static methods are defined using the `@staticmethod` decorator.
- in simple words, whenwe want to make the same value for all the objects of a class then we can use static method.

`example`
a IFC code is of static method as it need to be same for all the objects of the class.(all the customer usinf the same IFC code)

- Instance method is a method that belongs to an instance of a class. It has access to the instance (self) and can modify the instance's attributes. Instance methods are defined without any decorators.

`example`
a cgpa of a student is an instance method as it can be different for each student.
