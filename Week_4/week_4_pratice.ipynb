### Attributes

- Attributes are variable associated with a class/object. They define the state or characteristics of an object.

There are two types:

1. `Instance Attributes`: Unique to each object, defined inside the `__init__` method using self.
2. `Class Attributes`: Shared by all objects of a class, defined directly under the class block

### Methods

Methods are function defined inside a class. They define the behavior of an object

what it can do.
- The first parameter in instance methods is always self.
- You call them using `object.method_name()`


# Magic Methods or Dunder Methods

### what are Magic Methods?

Magic Methods are special built-in methods in Python that start and end with double iinderscores`(__like_this__)`. they are also known as dunder methids(shoert for "double undescore").

These methods are automatically called during specific operation like creating an object, printing it, adding two objects, comparing them, etc.

### Use of Magic methods

- to customize object behavior for built -in operations.
- to make your classes behave like built-in types(eg. how str() works on your object).
- to make debugging, logging, and operator overloading easier. 

```python
class book():
    def __init__(self, name_val, author_val):
        self.name = name_val
        self.author = author_val
    

    def my_method():
        print("this was called by the user\n")

my_book = book("LLM Fundemental", "Ankit")
print(my_book)

print("after applying dunder method\n")
### After appling the method
class book():
    def __init__(self, name_val, author_val):
        self.name = name_val
        self.author = author_val
    
    def __str__(self):  # Dunder method this to return when you say print
        return f"{self.name} has author {self.author}"

    def my_method():
        print("this was called by the user\n")

my_book = book("LLM Fundemental", "Ankit")
print(my_book)
```

```python
class money():
    def __init__(self, amount_val):
        self.amount = amount_val

    def __str__(self):
        return f"RS: {self.amount}/-"
    

wallet_1 = money(100)
print(wallet_1)
wallet_2 = money(400)

print(wallet_1 + wallet_2)
```

- you can see in above I can not add the number because it's not defined in the class I am going to define it below

```python
class money():
    def __init__(self, amount_val):
        self.amount = amount_val

    def __str__(self):
        return f"RS: {self.amount}/-"

    
    def __add__(self, other):
        return [self.amount + other.amount]  # now the number is printed in list
        # return money(self.amount + other.amount)  

wallet_1 = money(100)
wallet_2 = money(400)

print(wallet_1 + wallet_2)
```

- Now as you can see now it support the addition operation

### What is Encapsulation? 🔒

Encapsulation is the concept of hiding internal objects details and exposing only what is necessary. it helps in protecting data from unauthorized access and building data + methods that operate on the data in a single unit(i.e. a class).

|Syntax|Meaning
1. `name` --> Public attribute
2. `_name` --> Protected(internal use)
3. `__name` -->private(name mangling)


```python
class BankAccount:
    def __init__(self, owner, amount):
        self.owner = owner
        self.amount = amount

    def __str__(self):
        return f"Owner: {self.owner} has amount {self.amount}"
    
abc = BankAccount("Ankit", "1000")
abc.amount = 500000
print(abc)

```

```python
class BankAccount:
    def __init__(self, owner, amount):
        self.owner = owner
        self.amount = amount

    def __str__(self):
        return f"Owner: {self.owner} has amount {self.__amount}"
    
abc = BankAccount("Ankit", "1000")
abc.__amount = 500000
print(abc)

# when use __anyvar this tell that this is private data and do not need to ne changes
# name managgling abc_BankAccount__amount to acess it. 
```

```python
class BankAccount:
    def __init__(self, owner, amount):
        self.owner = owner
        self.__amount = float(amount)
        
    def deposit(self, amount):   ## getter 
        if amount > 0:
            self.__amount += amount
            print(f"Succesfully added {amount}")
        else:
            print("This was an invalid amount to added.")

    def withdraw(self, amount):    # getter
        if 0 < amount < self.__amount:
            self.__amount -= amount
            print(f"Succesfully withdrawn {amount}")
        else:
            print("This was an invalid amount to withdrawn")

    def get_balance(self):    #setter
        return self.__amount   


    def __str__(self):
        return f"Owner: {self.owner} has amount {self.__amount}"
    
abc = BankAccount("Ankit", "1000")
abc.deposit(5000)
print(abc.get_balance())
```
