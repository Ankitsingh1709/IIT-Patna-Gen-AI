```python
class Atm:
    def __init__(self):
        self.pin = ''
        self.balance = 0
        self.menu()
    
    def menu(self):
        while True:
            user_input = input('''
            ===============Welcome=============
            1. Press 1 to create pin\n
            2. Press 2 to check balance\n
            3. Press 3 to change pin\n
            4. Press 4 to withdraw \n
            5. Any thing else to exit
            ''')

            if user_input == '1':
                self.create_pin()
            elif user_input == '2':
                self.check_balance()
            elif user_input == '3':
                self.change_pin()
            elif user_input == '4':
                self.withdraw()
            # withdraw money
            else:
                print("Exiting... Bye!")
                break
            
    def create_pin(self):
        self.pin = input("Enter your pin:")
        self.balance = int(input("Enter your balance"))
        print("Pin created sucesfully")
        self.menu()
    
    def change_pin(self):
        old_pin = input("Enter your old pin: ")
        if old_pin == self.pin:
            new_pin = input("please enter your new pin")
            self.pin = new_pin
        else:
            print("Nhi krne dunga change 😂 wrong pin")
        self.menu()

    def check_balance(self):
        user_pin = input("Enter your pin:")
        if user_pin == self.pin:
            if self.balance < 1000:
                print("\nGareeb!\n Gareeb!\n Gareeb!\n")
                print(f"Your account balance is {self.balance}😶")
            else:
                print(f"Your account balance is {self.balance}🤑")
        else:
            print("Tobba tobba tobba pura mood krb kr diya 😶‍🌫️")
        self.menu()

    def withdraw(self):
        user_pin = input("Enter your pin:")
        if user_pin == self.pin:        
            user_input = int(input("Please enter the ammount to withdraw"))
            if user_input >= self.balance:
                print("Chl hatt garib! paise kaam daal 😣")
            else:
                self.balance -= user_input
                print(f"Withdraw successful\n Remaining balance: {self.balance}")
        else:
            print("Tobba tobba tobba pura mood krb kr diya 😶‍🌫️")
        self.menu()

    
obj = Atm()



```

### Write OOP classes to handel  the following scenarios:

1. A user can create and view 2D coordinates. 
2. A user can find out the distance between 2 coordinates.
3. A user can find the distance of a coordinate from origin.
4. A user can check if a point lies on a given line.
5. A user can find the distance btween a given 2D point and a given line. 

```python
class point:

    def __init__(self, x , y ):
        self.x = x
        self.y = y
        self.__str__()
        
    def __str__(self):
        return print(f"<{self.x},{self.y}>")
    
    def euclidean_distance(self, other):
        return ((self.x - other.x)**2 + (self.y - other.y)**2)**0.5
    
            
```

```python
p1 = point(0,0)
p2 = point(-2,-1)

p1.euclidean_distance(p2)
```
