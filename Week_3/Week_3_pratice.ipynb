
# Tuple

### what is a Tuple?

- A tuple is an ordered, immutable collection of values in Python, defined using parentheses() insted of square brackets [] which is for list

### why use Tuples?
- Faster than lists
- safe from accidental modification
- used for fixed data collection (coordinates, database records)
- Can be used as dictionary keys

```python
# tuple packing

abc = ()  # a emppty tuple
print(f"before packing{abc}")
num1 = 1
num2 = 2
str1 = "hello"

abc = num1, num2, str1

print(f"After packing{abc}")


# unpacking
new1, new2, new3 = abc

print("After unpacking\n")
print(new1)
print(new2)
print(new3)

```

```python
# Tuple Operations
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)  
tuple3 = tuple1 + tuple2  # Concatenation
print(f"Concatenated tuple: {tuple3}")
tuple4 = tuple1 * 2  # Repetition
print(f"Repeated tuple: {tuple4}")
# Membership Testing
print(f"Is 2 in tuple1? {'Yes' if 2 in tuple1 else 'No'}")
print(f"Is 5 in tuple1? {'Yes' if 5 in tuple1 else 'No'}")  
print(f"Length of tuple1: {len(tuple1)}")
print(f"Minimum of tuple1: {min(tuple1)}")
print(f"Maximum of tuple1: {max(tuple1)}")
```

```python
# slicing in tuple
my_tuple = (10, 20, 30, 40, 50)
print(f"Original tuple: {my_tuple}")
print(f"Sliced tuple (1:4): {my_tuple[1:4]}")  # index start from 0 in tuple the second number is where to stop 
# slicing but it is not included in the result
```

# Dictionary in Python

- A dictionary in Python is an unordered, mutable collection of key-value pairs.
- It is one of the most powerful built-in data structures, used to map unique keys to values. 

person = {"name": "Alice", "age": 25}

### why Use Dictionaries?

- Fast lookups by key
- keys are unique, values can be duplicated
- Great for structured data(like JSON, database rows, etc.)
- More readable and meaningful than lists for labeled data

```python
# Dict has no concept of indexing
# Dict is mutable dtype
# Dict key - immutable, value -mutable
# keys should be unique
```

```python
my_dict = {
    "name": "Alice",
    "age": 30,
    "city": "New York"
}
print(f"Original dictionary: {my_dict}")
# Accessing values
print(f"Name: {my_dict['name']}\n")       

# Adding a new key-value pair
my_dict["country"] = "USA"
print(f"Updated dictionary: {my_dict}\n")


# Modifying an existing value
my_dict["age"] = 31
print(f"Modified dictionary: {my_dict}|\n")


# Removing a key-value pair
del my_dict["city"]
print(f"Dictionary after deletion: {my_dict}\n")


# Dictionary Operations
dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3, "d": 4}
dict3 = {**dict1, **dict2}  # Merging dictionaries
print(f"Merged dictionary: {dict3}\n")


dict4 = {k: v * 4 for k, v in dict1.items()}  # Dictionary comprehension
print(f"Doubled values dictionary: {dict4}")
```

### Common Operations on  Dictionaries

```python
#Methods in dict
my_dict = {
    "name": "Alice",
    "age": 30,
    "city": "New York"
}
print(f"Original dictionary: {my_dict}\n")


# Using get() method
print(f"Name using get(): {my_dict.get('name')}\n")


# Using keys() method
print(f"Dictionary keys: {my_dict.keys()}\n") # gets the keys of the dict

# Using values() method
print(f"Dictionary values: {my_dict.values()}\n") # gets the values of the dict

# Using items() method
print(f"Dictionary items: {my_dict.items()}\n") # gets the key value pairs of the dict

# Using pop() method
removed_value = my_dict.pop("age") # removes the key value pair with the key "age" and returns the value
print(f"Removed value: {removed_value}\n")

# Using clear() method
my_dict.clear() # removes all the key value pairs from the dict 
print(f"Dictionary after clear: {my_dict}\n")

#using update() method
dict1 = {"a": 1, "b": 2}    
dict2 = {"c": 3, "d": 4}
dict1.update(dict2) # updates dict1 with the key value pairs from dict2
print(f"Updated dictionary: {dict1}\n") 
```

# What is a Set?

- A set is python is an unordered, unindexed collection of unique element . it is mutable, but elements must be immutable types (like strings, numbers, or tuples).

- Think of it like a mathematical set - no duplication allowed, and you can do operations like union, intresection, and difference. 

### when to use a set?
- When you need to store unique items (like a list of unique users)
- When you want to perform set operations (like finding common elements between two lists)  
- When you want to check for membership (like checking if an item is in a collection)
### Common Operations on Sets
- Adding elements: `my_set.add(1)`
- Removing elements: `my_set.remove(1)` 
- Union: `set1.union(set2)`
- Intersection: `set1.intersection(set2)`
- Difference: `set1.difference(set2)`

```python
# acessing set elements
my_set = {1, 2, 3, 4, 5}    
print(f"Original set: {my_set}\n")
print(f"Is 3 in the set? {'Yes' if 3 in my_set else 'No'}\n")  # Membership testing
# Adding an element to the set
my_set.add(6)   
print(f"Set after adding 6: {my_set}\n")
# Removing an element from the set  
my_set.remove(2)
print(f"Set after removing 2: {my_set}\n")


# Set Operations
set1 = {1, 2, 3}    
set2 = {3, 4, 5}

union_set = set1 | set2  # Union
print(f"Union of set1 and set2: {union_set}\n")

intersection_set = set1 & set2  # Intersection
print(f"Intersection of set1 and set2: {intersection_set}\n")

difference_set = set1 - set2  # Difference
print(f"Difference of set1 and set2: {difference_set}\n")

symmetric_difference_set = set1 ^ set2  # Symmetric Difference
print(f"Symmetric difference of set1 and set2: {symmetric_difference_set}\n") # elements that are in either set1 or set2 but not in both

```

### Methods of Set
- `add()`: Adds an element to the set.
- `remove()`: Removes a specific element from the set. Raises a KeyError if the
- `discard()`: Removes a specific element from the set. Does not raise an error if the element is not found.
- `clear()`: Removes all elements from the set.
- `union()`: Returns a new set that is the union of the current set and another set.
- `intersection()`: Returns a new set that is the intersection of the current set and another set.
- `difference()`: Returns a new set that is the difference between the current set and another set.
- `symmetric_difference()`: Returns a new set that is the symmetric difference between the current set and another set.
- `issubset()`: Checks if the current set is a subset of another set.
- `issuperset()`: Checks if the current set is a superset of another set.
- `copy()`: Returns a shallow copy of the set.
- `len()`: Returns the number of elements in the set.
- `in`: Checks if an element is in the set (e.g., `element in my_set`).
- `not in`: Checks if an element is not in the set (e.g., `element not in my_set`).


### Exception handling in Python
- Exception handling in Python is a mechanism to gracefully handle errors and exceptional conditions that may occur during the execution of a program. It allows developers to write code that can respond to errors without crashing the entire program.
### Why use Exception Handling?
- To prevent program crashes and maintain a good user experience.
- To handle anticipated errors (like file not found, division by zero, etc.) gracefully.
- To separate error-handling code from regular code, improving readability and maintainability.
### Common Exception Handling Constructs
- `try`: The block of code that may raise an exception is placed inside the `try` block.
- `except`: The block of code that handles the exception is placed inside the `except`          
block. You can specify the type of exception to catch (e.g., `except ValueError:`) or catch all exceptions with a generic `except:`.
- `else`: The block of code that executes if no exceptions were raised in the `try` block.
- `finally`: The block of code that executes regardless of whether an exception was raised or not. It is often used for cleanup actions (like closing files or releasing resources).    


```python
# code

try:
    members = int(input("Enter the number of members: "))
    result = 10 // members  # This will raise a ZeroDivisionError if members is 0
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")  
except ValueError:
    print("Error: Invalid input. Please enter a valid integer.")    
else:
    print(f"Result: {result}")
finally:
    print("This block will always execute. finally")  


# raise custom error
def calculate_area(radius):
    if radius < 0:
        raise ValueError("Radius cannot be negative.")
    return 3.14 * radius ** 2   
try:
    area = calculate_area(-5)  # This will raise a ValueError   
except ValueError as e:
    print(f"Error: {e}")
    
```
