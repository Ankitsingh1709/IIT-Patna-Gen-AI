### There are 2 types of file 

- Text file - all data is in text format and can be read by any text editor. Examples include .txt, .csv, .html, etc.
- Binary file - data is stored in a format that is not human-readable and requires specific software to read. Examples include .exe, .jpg, .pdf, etc.

How File I\O works in Python:
In Python, you can perform file I/O operations using built-in functions. The basic steps to work with files are:
1. Open the file using the `open()` function, which returns a file object.
2. Perform read or write operations on the file object. 
3. Close the file using the `close()` method or the `with` statement to ensure changes are saved and resources are released.

```python
# Write to a file

# Case 1 - if the file is not present

f = open('sample.txt', 'w')
f.write('Hello, this is a sample file.\n')
f.close() # Always close the file after writing to it to ensure that the data is saved properly.

# Once closed, you can write more in the file, if you want to add more data. need to reopen the file again


# write multiple lines in a file
f = open('sample1.txt', 'w')
f.write('Hello, this is a sample file.\n')
f.write('This is the second line.\n')
f.close()
```

```python
# case 2 - if the file is already present, it will overwrite the existing content of the file.
f= open("sample.txt", 'w')   
f.write('This is a new line.\n')
f.close()
```

- `w`  if you want to write the file and remove the existing content of the file.
- `a`  if you want to write the file and keep the existing content of the file. It will append the new content to the existing content of the file.

```python
# Reading from a file
# two type of functions to read a file
# 1. read() - it reads the entire content of the file and returns it as a string.
# 2. readline() - it reads a single line from the file and returns it as a string. It can be used in a loop to read the file line by line.

f = open('sample.txt', 'r')
content = f.read() # it reads the entire content of the file and returns it as a string.
f.close()
print(content)

# if you want to read small part

f = open('sample1.txt', 'r')
content = f.read(10) # it reads the entire content of the file and returns it as a string.
f.close()
print(content)
```

```python
# readline() - it reads a single line from the file and returns it as a string. It can be used in a loop to read the file line by line.

f = open('sample1.txt', 'r')
content = f.readline() # it reads a single line from the file and returns it as a string.
f.close()
print(content)
```

### Using Context Manager to handle files
Using a context manager (the `with` statement) is a best practice for handling files in Python. It ensures that the file is properly closed after its suite finishes, even if an exception is raised. This helps to prevent resource leaks and ensures that changes are saved.

```python
# with 

with open('sample.txt', 'w') as f:
    f.write('hehehe \n')
```

```python
# read 
with open('sample.txt', 'r') as f:
    print(f.read())
```

```python
# load files in chunks
big_file = ['hello world' for i in range(100)]

with open('big_file.txt', 'w') as f:
    f.writelines(big_file)


with open("big_file.txt", 'r') as f:

    chunk_size = 10

    while len(f.read(chunk_size)) > 0:
        print(f.read(chunk_size), end= '')
        f.read(chunk_size)
```
